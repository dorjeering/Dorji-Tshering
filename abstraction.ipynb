```python
#find out the distinction between attribute and properties

from abc import ABC, abstractmethod

class User(ABC):
  def __init__(self,name):
    self.name = name

  @abstractmethod
  def perform_action(self):
    pass

  def show_details(self):
    pass

class Customer(User):
  def perform_action(self):
    print(f'{self.name} is ordering food.')

  def show_details(self):
    print(f"Customer's Name: {self.name}")

class DeliveryPerson(User):
  def perform_action(self):
    print(f'{self.name} is delivering food.')

  def show_details(self):
    print(f"Delivery person's Name: {self.name}")

class Chef(User):
  def perform_action(self):
    print(f'{self.name} is preparing food.')

  def show_details(self):
    print(f"Chef's Name: {self.name}")

cus = Customer('Sonam')
deli = DeliveryPerson('Karma')
che = Chef('Tawla')

cus.perform_action()
cus.show_details()
print()
deli.perform_action()
deli.show_details()
print()
che.perform_action()
che.show_details()
```
